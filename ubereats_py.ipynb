```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/uberproject1/Uber_Eats_data.csv')
```

```python
df['rate'] = df['rate'].str.replace('/5', '', regex=False)
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')
df = df.dropna(subset=['rate'])
df['approx_cost(for two people)'] = df['approx_cost(for two people)'].str.replace(',', '', regex=False)
df['approx_cost(for two people)'] = pd.to_numeric(df['approx_cost(for two people)'], errors='coerce')

print("Dataset reloaded and specified columns cleaned:")
df.info()
df.head()
```

```python
df = df.drop(columns=['phone', 'listed_in(city)'])
df.head()
```

```python
print(f"Number of duplicate rows before removal: {df.duplicated().sum()}")
df = df.drop_duplicates(subset=None, keep='first')
print(f"DataFrame shape after duplicate removal: {df.shape}")
```

```python
df = df.rename(columns={
    'approx_cost(for two people)': 'approx_cost_fortwo',
    'listed_in(type)': 'restaurant_type'
})
df.head()
```

```python
print(f"DataFrame Shape: {df.shape}")
print(f"DataFrame Columns: {df.columns.tolist()}")
print(f"\nDataFrame Data Types:\n{df.dtypes}")
print(f"\nNull Values per Column:\n{df.isnull().sum()}")
print(f"\nDataFrame Head:\n{df.head()}")
```

```python
import numpy as np

grouped = df.groupby('name').agg(
    avg_rating=('rate', 'mean'),
    total_votes=('votes', 'sum')
)

grouped = grouped[grouped['total_votes'] > 50]

grouped['score'] = grouped['avg_rating'] * np.log1p(grouped['total_votes'])

top_restaurants = grouped.sort_values(by='score', ascending=False)

top_restaurants.head(10)
```

```python
import pandas as pd

# Query 1 - Highest rated locations
q1 = """
SELECT location, ROUND(AVG(rate),2) as avg_rating
FROM restaurants
GROUP BY location
ORDER BY avg_rating DESC
LIMIT 10;
"""

# Query 2 - Most restaurants per location
q2 = """
SELECT location, COUNT(name) as total_restaurants
FROM restaurants
GROUP BY location
ORDER BY total_restaurants DESC
LIMIT 10;
"""

# Run and display
df_q1 = pd.read_sql_query(q1, conn)
df_q2 = pd.read_sql_query(q2, conn)

print("Q1 - Top Rated Locations:")
print(df_q1)
print("\nQ2 - Most Restaurants:")
print(df_q2)
```

```python
# .How do low, mid, and premium-priced restaurants perform in terms of ratings?

q5="""SELECT
    CASE
        WHEN approx_cost_fortwo < 500 THEN 'Low'
        WHEN approx_cost_fortwo between 500 AND 800 THEN 'Mid'
        ELSE 'Premium'
    END as price_segment,
    AVG(rate) as avg_rating,
    COUNT(name) as total_restaurants
FROM restaurants
GROUP BY price_segment
ORDER BY avg_rating DESC;"""
df_q5 = pd.read_sql_query(q5, conn)
print("Q5 - Price Segments:")
print(df_q5)
```

```python
 # Which cuisines are most common in Bangalore?

# 6.Step 1 - Python handles the split
cuisine_df = df['cuisines'].str.split(',').explode().str.strip()

# Step 2 - Count using Python
cuisine_counts = cuisine_df.value_counts().head(10)
print(cuisine_counts)
```

```python
# It still tells us:
# Which cuisine CATEGORY tends to appear in
# higher rated restaurants!
#7."Cuisine ratings are approximated by assigning restaurant ratings to each cuisine served — this is a known limitation as individual cuisine performance data is unavailable"
cuisine_rating = df.assign(cuisine=df['cuisines'].str.split(',')).explode('cuisine')
cuisine_rating['cuisine'] = cuisine_rating['cuisine'].str.strip()
q7_result = cuisine_rating.groupby('cuisine')['rate'].mean().sort_values(ascending=False).head(10)
print(q7_result)
```

8.Which cuisines perform well despite having fewer restaurants?(9 in pdf)

```python
cuisine_summary = cuisine_rating.groupby('cuisine').agg(
    avg_rating=('rate', 'mean'),
    total_count=('rate', 'count')
).reset_index()

# Filter: HIGH rating but LOW count
q8_result = cuisine_summary[
    (cuisine_summary['avg_rating'] > 4.0) &
    (cuisine_summary['total_count'] < 500)
].sort_values('avg_rating', ascending=False).head(10)

print(q8_result)
```

```python
q9= """
SELECT
    CASE
        WHEN approx_cost_fortwo < 500 THEN 'Low'
        WHEN approx_cost_fortwo BETWEEN 500 AND 800 THEN 'Mid'
        ELSE 'Premium'
    END as cost_category,
    ROUND(AVG(rate), 2) as avg_rating,
    ROUND(AVG(approx_cost_fortwo), 2) as avg_cost,
    COUNT(name) as total_restaurants
FROM restaurants
GROUP BY cost_category
ORDER BY avg_cost ASC;
"""

# Python executes it against SQLite!
df_q9 = pd.read_sql_query(q9, conn)
print("Q9 - Cost vs Rating:")
print(df_q9)
```

```python
#"Which locations are ideal for premium restaurant onboarding?"
q11 = """
SELECT
    location,
    ROUND(AVG(rate), 2) as avg_rating,
    ROUND(AVG(approx_cost_fortwo), 2) as avg_cost,
    ROUND(AVG(votes), 0) as avg_votes,
    COUNT(name) as total_restaurants
FROM restaurants
GROUP BY location
HAVING AVG(rate) > 4.0
AND AVG(approx_cost_fortwo) > 800
AND COUNT(name) > 10
ORDER BY avg_rating DESC, avg_cost DESC
LIMIT 10;
"""
df_q11 = pd.read_sql_query(q11, conn)
print("Q11 - Premium Locations:")
print(df_q11)
```

```python
# Which restaurants are top performers within each pricing segment?

# Step 1 - Calculate score in Python
scored_df = df.groupby(['name', 'location',
                         'cuisines',
                         'approx_cost_fortwo']).agg(
    avg_rating=('rate', 'mean'),
    total_votes=('votes', 'sum')
).reset_index()

scored_df = scored_df[scored_df['total_votes'] > 50]
scored_df['score'] = round(
    scored_df['avg_rating'] * np.log1p(scored_df['total_votes']), 2
)

# Step 2 - Push to SQLite
scored_df.to_sql('restaurant_scores', conn,
                  if_exists='replace', index=False)

# Step 3 - Query top performers per price segment
 # LOW segment - given to you as example!
q15_low = """
SELECT
    'Low' as price_segment,
    name, location, avg_rating, total_votes, score
FROM restaurant_scores
WHERE approx_cost_fortwo < 500
ORDER BY score DESC
LIMIT 5;
"""
df_q15_low = pd.read_sql_query(q15_low, conn)
print("Q15 - Top Performers Per Segment:")
print(df_q15_low)
```

```python
# MID segment - YOU complete!
q15_mid = """
SELECT
    'MID' as price_segment,
    name, location, avg_rating, total_votes, score
FROM restaurant_scores
WHERE approx_cost_fortwo >= 500
AND approx_cost_fortwo < 800
ORDER BY score DESC
LIMIT 5;
"""

# PREMIUM segment - YOU complete!
q15_premium = """
SELECT
    'PREMIUM' as price_segment,
    name, location, avg_rating, total_votes, score
FROM restaurant_scores
WHERE approx_cost_fortwo >= 800
ORDER BY score DESC
LIMIT 5;
"""
# Combine into one DataFrame!
df_q15_final = pd.concat([
    pd.read_sql_query(q15_low, conn),
    pd.read_sql_query(q15_mid, conn),
    pd.read_sql_query(q15_premium, conn)
])
print("Q15 - Top 5 Per Segment:")
print(df_q15_final)
```
